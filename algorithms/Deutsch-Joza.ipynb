## Deutsch/Deutsch-Joza Algorithm

### Introduction

The Deutsch-Joza algorithm is a foundational quantum algorithm developed by David Deutsch and Richard Joza as an expansion on Deutsch's original algorithm, which only interacted with a two qubit state. Deutsch's original algorithm was originally designed wiht the input of 1-qubit and provided a 1-qubit output. However, he later expanded on this idea with his partner Richard Joza to be able to accept n qubits in 1992. 

Deutsch's Algorithm makes use of a toy function that can take either 0 or 1 as an input, and detect if the function was balanced or constant. 

| Function | $x=0$   | $x = 1$ |
|:--------:|:-------:|:-------:|
| $f_1$    |    0    |    0    |
| $f_2$    |    0    |    1    |
| $f_3$    |    1    |    0    |
| $f_4$    |    1    |    1    |

In the table above $f_1$ and $f_4$ are constant functions because the output is the same for $x=0$ and $x=1$. Furthermore, $f_2$ and $f_3$ are considered balanced because the inputs $x=0$ and $x=1$ give different outputs. On a classical machine, this problem would require evaluating both values of $x$ and examining the results to determine whether the function was constant like $f_{1,4}$, or balanced like $f_{2,3}$, hence Deutch's problem is analogous to determining whether or not a coin is fair or not. However, on a quantum computer we can solve this problem with just one measurement, providing a signifcant advantage over classical machines.

Although this problem is trivial and does not provide any practical applications, Deutch's algorithm was extremely significant because it was the first algorithm that demonstrated a quantum advantage over classical machines. Because of this, it is credited with starting the field of quantum computing.
